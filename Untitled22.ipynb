```python
# japanize-matplotlibのインストール
!pip install japanize-matplotlib -qq

import matplotlib.pyplot as plt
import japanize_matplotlib  # 日本語化を有効にする
import matplotlib.dates as mdates  # 日付フォーマット用

# マイナス記号を正しく表示
plt.rcParams['axes.unicode_minus'] = False

import requests
import pandas as pd
import numpy as np
from bs4 import BeautifulSoup
from io import StringIO

# ======= 地域設定を変数化 =======
location = {
    "name": "稲取",
    "prec_no": "50",
    "block_no": "0457"
}

year = 2025
months = [2, 3]

# DTS法の計算関数（平均気温から指数計算）
def calculate_dts(avg_temp):
    t_kelvin = avg_temp + 273.15  # typo修正: 27715 → 273.15
    return np.exp((9.5 * 10**3 * (t_kelvin - 288.2)) / (t_kelvin * 288.2))

all_data = pd.DataFrame()

for month in months:
    url = (f"https://www.data.jma.go.jp/stats/etrn/view/daily_a1.php?"
           f"prec_no={location['prec_no']}&block_no={location['block_no']}&year={year}&month={month}&day=&view=p1")

    print(f"{year}年{month}月のデータを取得中... ({location['name']})")

    response = requests.get(url)
    response.encoding = 'UTF-8'

    soup = BeautifulSoup(response.text, 'html.parser')

    table = soup.find('table', id='tablefix1')
    if table is None:
        print("指定されたテーブル（id='tablefix1'）が見つかりませんでした。")
        continue
    else:
        df_list = pd.read_html(StringIO(str(table)), header=0)
        if len(df_list) == 0:
            print("テーブルからデータを抽出できませんでした。")
            continue
        df = df_list[0]

    df_selected = df.iloc[:, [0, 1, 4, 5]]
    df_selected.columns = ['day', 'precip', 'avg_temp', 'max_temp']

    df_selected = df_selected[df_selected['day'].apply(lambda x: str(x).strip().isdigit())]
    df_selected = df_selected.reset_index(drop=True)

    df_selected['date'] = pd.to_datetime(
        df_selected['day'].apply(lambda d: f"{year}-{month}-{int(d)}")
    )

    df_selected = df_selected[['date', 'precip', 'avg_temp', 'max_temp']]

    all_data = pd.concat([all_data, df_selected], ignore_index=True)

# 数値変換
all_data["avg_temp"] = pd.to_numeric(all_data["avg_temp"], errors='coerce')
all_data["max_temp"] = pd.to_numeric(all_data["max_temp"], errors='coerce')
all_data["precip"]   = pd.to_numeric(all_data["precip"], errors='coerce')
all_data = all_data.dropna().reset_index(drop=True)

# 累積計算
all_data["avg_cum"] = all_data["avg_temp"].cumsum()
all_data["max_cum"] = all_data["max_temp"].cumsum()
all_data["dts_daily"] = all_data["avg_temp"].apply(calculate_dts)
all_data["dts_cum"] = all_data["dts_daily"].cumsum()

# 開花予測基準
avg_target = 400   # 400度法則（平均気温）
max_target = 600   # 600度法則（最高気温）
dts_target = 27.53 # DTS法

all_data["avg_pct"] = (all_data["avg_cum"] / avg_target) * 100
all_data["max_pct"] = (all_data["max_cum"] / max_target) * 100
all_data["dts_pct"] = (all_data["dts_cum"] / dts_target) * 100

# 開花日算出
avg_bloom_day = all_data.loc[all_data["avg_cum"] >= avg_target, "date"].iloc[0]
max_bloom_day = all_data.loc[all_data["max_cum"] >= max_target, "date"].iloc[0]
dts_bloom_day = all_data.loc[all_data["dts_cum"] >= dts_target, "date"].iloc[0]

print(f"\n🌸 {location['name']}ソメイヨシノ 開花予想日 🌸")
print(f"400度法則（平均気温）: {avg_bloom_day.date()}")
print(f"600度法則（最高気温）: {max_bloom_day.date()}")
print(f"DTS法: {dts_bloom_day.date()}")

# --- グラフ作成 ---
fig, ax1 = plt.subplots(figsize=(12, 7))
ax2 = ax1.twinx()

# 左軸：開花予測達成率
ax1.plot(all_data["date"], all_data["avg_pct"], label="400度法則 (平均気温)", linewidth=2, color="red")
ax1.plot(all_data["date"], all_data["max_pct"], label="600度法則 (最高気温)", linewidth=2, color="green")
ax1.plot(all_data["date"], all_data["dts_pct"], label="DTS法", linewidth=2, color="orange")
ax1.axhline(100, color='gray', linestyle='--', linewidth=1)  # 100%線
ax1.axhline(0, color='black', linestyle='-', linewidth=2)    # ゼロを太い実線で追加

# 軸ラベルとタイトル
ax1.set_xlabel("日付（2月1日～）")
ax1.set_ylabel("開花予測達成率 (%)")
ax1.set_title(f"{location['name']} ソメイヨシノ 開花予想 累積曲線", fontsize=16)

# 右軸：温度と日雨量
ax2.plot(all_data["date"], all_data["avg_temp"], color="blue", linewidth=1, label="平均気温")
ax2.plot(all_data["date"], all_data["max_temp"], color="blue", linewidth=1, label="最高気温")
ax2.bar(all_data["date"], all_data["precip"], width=0.5, color="black", alpha=0.5, label="日雨量")
ax2.set_ylabel("温度 (°C) / 降水量 (mm)")

# 日付をわかりやすくするグリッド設定
ax1.xaxis.set_major_locator(mdates.DayLocator(interval=5))  # 5日ごとに主目盛り
ax1.xaxis.set_minor_locator(mdates.DayLocator(interval=1))  # 1日ごとに副目盛り
ax1.xaxis.set_major_formatter(mdates.DateFormatter('%m/%d'))  # 月/日形式
plt.setp(ax1.get_xticklabels(), rotation=45, ha='right')  # ラベルを45度回転

# メッシュ（グリッド）を背景に
ax1.grid(which='major', linestyle='-', linewidth=0.5, alpha=0.7)  # 主グリッド
ax1.grid(which='minor', linestyle=':', linewidth=0.3, alpha=0.5)  # 副グリッド

# 凡例
lines1, labels1 = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
ax1.legend(lines1 + lines2, labels1 + labels2, loc="upper left")

ax2.set_ylim(ax1.get_ylim())  # 両軸のスケールを揃える
plt.tight_layout()
plt.show()
```
